```python
import torch
import numpy as np
#convert sparse matrix to dense
X_train_array=X_train_preprocessed.toarray().astype(np.float32)
X_val_array=X_val_preprocessed.toarray().astype(np.float32)
X_test_array=X_test_preprocessed.toarray().astype(np.float32)

#convert features to tensors
X_train_tensor=torch.from_numpy(X_train_array)
X_val_tensor=torch.from_numpy(X_val_array)
X_test_tensor=torch.from_numpy(X_test_array)


#convert targets to tensors
y_train_tensor=torch.tensor(y_train.to_numpy(),dtype=torch.float32).reshape(-1,1)
y_val_tensor=torch.tensor(y_val.to_numpy(),dtype=torch.float32).reshape(-1,1)
y_test_tensor=torch.tensor(y_test.to_numpy(),dtype=torch.float32).reshape(-1,1)

```

```python
print("training features:",X_train_tensor.shape)
print("validation features:",X_val_tensor.shape)
print("testing features:",X_test_tensor.shape)

print("training target:",y_train_tensor.shape)
print("validation target:",y_val_tensor.shape)
print("testing target:",y_test_tensor.shape)

print("features data type:",X_train_tensor.dtype)
print("targets data type:",y_train_tensor.dtype)
```

```python
#linear regression
import torch.nn as nn
linear_model=nn.Sequential(nn.Linear(X_train_tensor.shape[1],1))
loss_fn=nn.MSELoss()
optimizer=torch.optim.Adam(linear_model.parameters(),lr=0.001)
print(linear_model)
```

```python
#deep nural network
deep_model=nn.Sequential(
    nn.Linear(X_train_tensor.shape[1],64),
    nn.ReLU(),
    nn.Linear(64,64),
    nn.ReLU(),
    nn.Linear(64,64),
    nn.ReLU(),
    nn.Linear(64,1)
)
loss_fn=nn.MSELoss()
optimizer=torch.optim.Adam(deep_model.parameters(),lr=0.001)
print(deep_model)

```

```python

def build_model(hidden_layers):
    layers = []
    input_size = X_train_tensor.shape[1]

    for neurons in hidden_layers:
        layers.append(nn.Linear(input_size, neurons))
        layers.append(nn.ReLU())
        input_size = neurons

    layers.append(nn.Linear(input_size, 1))

    return nn.Sequential(*layers)
```

```python
model = build_model([64, 64, 64])
print(model)
```

```python

from torch.utils.data import TensorDataset, DataLoader

def train_and_evaluate(model, epochs=30, batch_size=64):

    loss_fn = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

    train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        model.train()

        for X_batch, y_batch in train_loader:

            predictions = model(X_batch)
            loss = loss_fn(predictions, y_batch)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        model.eval()

        with torch.no_grad():
            train_predictions = model(X_train_tensor)
            val_predictions = model(X_val_tensor)

            train_mse = loss_fn(
                train_predictions, y_train_tensor
            ).item()

            val_mse = loss_fn(
                val_predictions, y_val_tensor
            ).item()

        train_losses.append(train_mse)
        val_losses.append(val_mse)

    return train_losses, val_losses
```

```python
#exp 1
import time

model_1 = build_model([32, 32, 32])

start_time = time.perf_counter()

train_losses_1, val_losses_1 = train_and_evaluate(model_1)

experiment_time_1 = time.perf_counter() - start_time

best_epoch_1 = np.argmin(val_losses_1)

print("Architecture: [32, 32, 32]")
print("Best Training MSE:", train_losses_1[best_epoch_1])
print("Best Validation MSE:", val_losses_1[best_epoch_1])
print("Best Epoch:", best_epoch_1 + 1)
print("Training Time:", round(experiment_time_1, 2), "seconds")
```

```python

# Experiment 15 - Huber Loss

def train_and_evaluate_huber(model, epochs=30, batch_size=64):

    loss_fn = nn.HuberLoss(delta=1.0)
    mse_fn = nn.MSELoss()

    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

    train_dataset = TensorDataset(X_train_tensor, y_train_tensor)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        model.train()

        for X_batch, y_batch in train_loader:

            predictions = model(X_batch)
            loss = loss_fn(predictions, y_batch)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        model.eval()

        with torch.no_grad():

            train_predictions = model(X_train_tensor)
            val_predictions = model(X_val_tensor)

            train_mse = mse_fn(
                train_predictions, y_train_tensor
            ).item()

            val_mse = mse_fn(
                val_predictions, y_val_tensor
            ).item()

        train_losses.append(train_mse)
        val_losses.append(val_mse)

    return train_losses, val_losses


model_15 = build_model([64, 64, 64])

start_time = time.perf_counter()

train_losses_15, val_losses_15 = train_and_evaluate_huber(model_15)

experiment_time_15 = time.perf_counter() - start_time

best_epoch_15 = np.argmin(val_losses_15)

print("Loss Function: Huber")
print("Architecture: [64, 64, 64]")
print("Best Training MSE:", train_losses_15[best_epoch_15])
print("Best Validation MSE:", val_losses_15[best_epoch_15])
print("Best Epoch:", best_epoch_15 + 1)
print("Training Time:", round(experiment_time_15, 2), "seconds")
```
